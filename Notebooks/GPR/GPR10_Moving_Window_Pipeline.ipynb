# Pipeline Moving Window
- Here, we perform moving window analysis using the pipeline code fit at various datasplits.

```python
# Importing and data hopo hopo
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import sys
from pathlib import Path

sys.path.append(str(Path('../..').resolve()))
from helpers.gpr import train_gpr
```

```python
def plot_moving_window( n_windows = 10, train_low = 0.1, train_high = 0.9):
    fig, ax = plt.subplots(n_windows, 1, figsize=(20, 5 * n_windows), sharex= True, sharey= True)
    train_splits = np.linspace(train_low, train_high, n_windows)       

    for idx, train_split in enumerate(train_splits):
        print(f"{idx+1}/{n_windows}")
        valid_split = 0.99 - train_split
        best_gp = train_gpr(
        datapath    =  r"../../Data/benchmark/HD201091_Mt_wilson_data.txt",
        train_split=train_split,
        valid_split=valid_split,
        star_name   = 'HD201091',
        star_type   = 'G',
        add_prefix  = False,
        relative=True,
        verbose          = False,
        plot             = False,
        loop_verbose     = False,
        loop_plot        = True,
        loop_savefigs    = True,   # show inline rather than saving
        results_verbose  = True,
        results_plot     = True,
        require_mid=True,
        ax = ax[idx]
    )
    
    plt.savefig("moving_window_HD201091")

plot_moving_window()
```
